## Exercise 1: Number Guessing Game

```python
import random
from typing import Any


def guessing_game():
    
    number: int = random.randint(a=0, b=100)
    
    user_input = 0
    
    while user_input != number:
    
        user_input: Any = eval(input("Enter your guess:"))
        
        if user_input > number:
            print("Too High")
        elif user_input < number:
            print("Too Low")
        elif user_input == number:
            print("Just Right")

if __name__ == "__main__":
    guessing_game()
```
